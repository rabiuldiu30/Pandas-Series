Importing Pandas

```python
import numpy as np
import pandas as pd
```

Series from lists

```python
# string
country = ['Bangladesh','Pakisthan','Turkey','Nepal']

pd.Series(country)
```

```python
# integers
runs = [12,34,9,89,100]

pd.Series(runs)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','computer']

pd.Series(marks,index=subjects)
```

```python
# setting a name

pd.Series(marks,index=subjects, name="Rabiul's marks")
```
